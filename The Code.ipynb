```python
from  selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import time

options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
#options.add_argument("--headless=new")
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--disable-notifications")
options.add_argument("--disable-popup-blocking")
driver = webdriver.Chrome (options=options)
driver.get('https://wuzzuf.net/')

positions=['Data engineer','Data Science' ,'AI Developer','Machine Learning Engineer']
counter=0
jobs_info=[]
pages=3
for i in positions:
    time.sleep(5)
    driver.find_element(By.NAME,'q').send_keys(i , Keys.ENTER)
    time.sleep(2)
    driver.refresh()
    time.sleep(4)
    for j in range(pages):
        jobs=driver.find_elements(By.CLASS_NAME,'css-pkv5jc')
        time.sleep(5)
        for job in jobs:
            
            try:
                job_number= counter
            except:
                job_number= 'there isnot'
            try:
                job_title=job.find_element(By.CLASS_NAME,'css-o171kl').text
            except:
                job_title='there isnot'
            try:
                job_href=job.find_element(By.CLASS_NAME,'css-o171kl').get_attribute('href')
            except:
                job_href='there isnot'

            try:
                job_company=job.find_element(By.CLASS_NAME,'css-ipsyv7').text
            except:
                job_company='there isnot'
            
            try:
                job_location=job.find_element(By.CLASS_NAME,'css-16x61xq').text
            except:
                job_location='there isnot'

            try:
                job_type= job.find_element(By.CLASS_NAME,'css-5jhz9n').text. replace ('\n','/')
            except:
                job_type= 'there isnot'

            try:
                job_exp=job.find_element(By.CSS_SELECTOR,'div > div.css-1rhj4yg > div:nth-child(2) > span').text
            except:
                job_exp='there isnot'
            
        
            counter+=1
            jobs_info.append(
            {
                'job number': job_number,
                'job_title':job_title,
                'job_href':job_href,
                'job_company':job_company,
                'job_location':job_location,
                'job_type':job_type,
                'job_exp':job_exp,
                'job_category': i
            }
            )
        try:
            driver.find_element(By.XPATH, "//a[@aria-label='Next page']//i[@class='css-b80bsd efou2fk0']//*[name()='svg']").click()
            time.sleep(3)
        except:
            print('problem')
            break

    time.sleep(5)
```
